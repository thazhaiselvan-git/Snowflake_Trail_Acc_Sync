```
import copy
orig = [[1, 2], [3, 4]]
sh = copy.copy(orig)      # new outer list, same inner lists
dp = copy.deepcopy(orig)  # everything copied recursively
orig[0][0] = 99
sh[0][0]  # 99  (shared)
dp[0][0]  # 1

print (orig)
print (sh)
print (dp)
```

```
# *args collects extra positional arguments into a tuple
def add_all(*args):
    print("args:", args)
    return sum(args)

print(add_all(1, 2, 3))        # args: (1, 2, 3) -> 6

# **kwargs collects extra keyword arguments into a dict
def show_profile(**kwargs):
    print("kwargs:", kwargs)
    for key, value in kwargs.items():
        print(f"{key} = {value}")

show_profile(name="Thazhai", role="Data Engineer", exp=3)

# Combining them (order: regular, *args, keyword-only, **kwargs)
def order_summary(customer, *items, discount=0, **details):
    print("Customer:", customer)
    print("Items:", items)
    print("Discount:", discount)
    print("Details:", details)

order_summary("Alice", "pen", "book", discount=10, city="Chennai", paid=True)

# Unpacking: * and ** also spread values into a call
nums = [4, 5, 6]
info = {"name": "Bob", "role": "Analyst"}
print(add_all(*nums))          # same as add_all(4, 5, 6)
show_profile(**info)           # same as show_profile(name="Bob", role="Analyst")

# Common interview use: a wrapper that passes everything through
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with {args} {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@logger
def multiply(a, b):
    return a * b

print(multiply(3, b=4))        # Calling multiply with (3,) {'b': 4} -> 12

@logger
def add(a, b,c,d):
    return a + b + c + d
print(add(1,2,c=3, d=4)) 
```

```
a = [1, 2, 3]
a.append([4, 5])     # adds ONE item (the whole list)
print(a)             # [1, 2, 3, [4, 5]]  -> len 4

b = [1, 2, 3]
b.extend([4, 5])     # adds EACH item from the iterable
print(b)             # [1, 2, 3, 4, 5]    -> len 5

# With a string
x = ['a']; x.append('hi');  print(x)   # ['a', 'hi']
y = ['a']; y.extend('hi');  print(y)   # ['a', 'h', 'i']
```

```
s = " Snowflake Data Cloud "

# 1. strip(): remove whitespace at the start and end (also lstrip / rstrip)
print((s.strip()))         # 'Snowflake Data Cloud'
print(repr(s.lstrip()))         # 'Snowflake Data Cloud '
print(repr(s.rstrip()))         # ' Snowflake Data Cloud'
print("xxSnowxx".strip("x"))    # 'Snow' (strips the given characters)

# 2. lower() / upper() (also title, capitalize, swapcase)
print(s.lower())                # ' snowflake data cloud '
print(s.upper())                # ' SNOWFLAKE DATA CLOUD '
print("data cloud".title())    # 'Data Cloud'

# 3. split(): string -> list
print(s.split())                # ['Snowflake', 'Data', 'Cloud'] (no argument: splits on any whitespace and drops the extra spaces)
print("a,b,,c".split(","))      # ['a', 'b', '', 'c']
print("a,b,c,d,e".split(",", 1))    # ['a', 'b,c'] (at most 1 split)

# 4. join(): list -> string (the opposite of split)
print("-".join(["a", "b"]))     # 'a-b'
words = s.split()
print("_".join(words))
print("_".join(words).lower())  # 'snowflake_data_cloud'

# 5. replace()
print(s.replace("Data", "AI"))  # ' Snowflake AI Cloud '
print("aaa".replace("a", "b", 2))  # 'bba' (replace t most 2)

# 6. find() vs index()
print(s.find("Data"))           # 11
print(s.find("n")) 
print(s.find("SQL"))            # -1 (not found)
# s.index("SQL")                # raises ValueError instead of returning -1
print("Data" in s)              # True (simplest way to check)
print("SQL" in s) 
# 7. startswith() / endswith()print(s.startswith(" Snow"))    # Trueprint(s.strip().endswith("Cloud"))  # Trueprint("file.csv".endswith((".csv", ".txt")))  # True (a tuple checks several options)
# 8. f-string formatting
pi, n, name = 3.14159, 1234567, "Snowflake"
print(f"{pi:.2f}")              # '3.14'
print(f"{n:,}")                 # '1,234,567'
print(f"{0.256:.1%}")           # '25.6%'
print(f"[{name:>12}]")          # '[   Snowflake]' (right-align to width 12)
print(f"{name=}")               # "name='Snowflake'" (handy for debugging)

# 9. Slicing and reversing
t = s.strip()
print(s[::-1])                  # ' duolC ataD ekalfwonS '
print(s[1:5])
print(s[1:5:10])
#print(s[::5])  
#print(t[0::])  
print(t[:9])                    # 'Snowflake'
print(t[-5:])                   # 'Cloud'
print(t[::2])                   # 'SofaeDt lu' (every 2nd character)

## Strings are immutable: these methods return a NEW string
#s.upper()
#print(s)                        # unchanged: ' Snowflake Data Cloud '
#s = s.strip()                   # reassign to keep the result
#
## Common interview problems
#print(t == t[::-1])                      # palindrome check -> False
#print(" ".join(w[::-1] for w in t.split()))  # 'ekalfwonS ataD duolC'
#print(t.count("a"))                      # 3
```
